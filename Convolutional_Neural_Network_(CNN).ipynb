# Building a Convolutional Neural Network (CNN) using TensorFlow & Keras.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
```

### Task 1:Build the CNN Architecture


```python
def generate_model():
    model = tf.keras.Sequential([
        tf.keras.Input(shape=(28, 28, 1)),

        # first layer :
        layers.Conv2D(
            16,  # filters (future map)
            (3, 3), # kernel_size
           activation='relu',
         input_shape=(28, 28, 1)
          ),
        # 2nd layer: MaxPool2D : reduces the spatial dimensions of the feature maps by selecting the maximum value in each 2×2 region
        layers.MaxPool2D(
          pool_size=(2, 2), ),

        # The second convolutional block
        layers.Conv2D(32, (3, 3), activation='relu'),
        layers.MaxPool2D(pool_size=(2, 2), strides=2),


        #  Fully connected classifier
        # flatten layer ==> cinvert to one vector
        layers.Flatten(),
        # dense layer
        layers.Dense(128, activation='relu'),
        # outpyr layer :it shows 10 features woth sum withim 100 using softmax
        layers.Dense(10, activation='softmax')

    ])
    return model
```

```python
model = generate_model()
model.summary()
```

- The model consists of two convolutional layers with 16 and 32 filters, respectively.
- Each convolutional layer uses a 3×3 kernel and ReLU activation, followed by a 2×2 MaxPooling layer to reduce the spatial dimensions.
- The extracted feature maps are converted into a one-dimensional vector using Flatten.
- A Dense layer with 128 neurons processes these features, followed by an output layer with 10 neurons and Softmax activation for multi-class classification.

### Task 2 : Modify the CNN Model

```python
def generate_model_modified():
  model = tf.keras.Sequential([
    tf.keras.Input(shape=(28, 28, 1)),

    # first conv block
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),

    # second conv block
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),

    # fully connected classifer
    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dense(10, activation='softmax')
         ])

  return model
```

```python
model_2 = generate_model_modified()
model_2.summary()
```

* note:

in both models , the hight and width of future maps didn't change, as we haven't changge teh kernel size , stride , padding ; what changes is the number of the channel, so it change the number of parampeters after flatten , and this because of the filters, dense neurons

```python

# Compare the total parameters of both CNN models

params_model = model.count_params()
params_model_2 = model_2.count_params()

print("Model 1 Parameters:", params_model)
print("Model 2 Parameters:", params_model_2)

difference = params_model_2 - params_model

print("Parameter Difference:", difference)
print("Increase Ratio:", round(params_model_2 / params_model, 2))
```

- The number of parameters increased because we increased the number of filters in both convolutional layers and the number of neurons in the Dense layer.

- Increasing the filters allows the model to learn more feature maps. It also increases the number of input features passed to the Dense layer after Flatten, resulting in more trainable weights.

- The modified model has a higher learning capacity, but it also requires more computational resources and may be more prone to overfitting.

### Task 3 — Compile the Model

```python

# Task 3: Compile the modified CNN model

model_2.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully!")
```

The CNN model was compiled using:

- Adam Optimizer: Updates weights via adaptive learning rates to minimize loss.
- Sparse Categorical Crossentropy: Measures error between predicted probabilities and integer-encoded true labels for multi-class classification.
- Accuracy Metric: Tracks the proportion of correctly classified samples during training and evaluation.
